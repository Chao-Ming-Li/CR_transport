# Proton transport and energy losses
Analyze `transport_loss.cpp`: 10,000 protons injected at the origin, logarithmically spaced initial total energies, constant hydrogen density, linear vertical wind, adiabatic and continuous pp losses, and no escape boundary.

Run this notebook from `output/` or the project directory. Both files must come from the same run; they contain no parameter metadata.
- Final records: `(x, y, z, E)`, native-endian float64, positions in pc and total proton energy in GeV.
- Solar-ring records: `(E, time_yr)`, native-endian float64, total energy in GeV and elapsed propagation time in years.

Every step ending inside the solar ring is recorded, including repeat samples of the same proton. Records are grouped by particle, not sorted by time; no particle IDs are saved.


```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Parameters and filenames must match the C++ run.
data_dir = Path('.')
if not (data_dir / 'proton_nH0.1_t1e9.bin').exists():
    data_dir = Path('output')
file_final = data_dir / 'proton_nH0.1_t1e9.bin'
file_earth = data_dir / 'proton_earth_nH0.1_t1e9.bin'
E_min, E_max = 10.0, 1e6  # initial total energy, GeV
expected_particles = 10000
t_end = 1e9              # yr
nH = 0.1                 # cm^-3
a = 1e-8                 # div(v), yr^-1
m_p = 0.938272           # proton rest energy, GeV
c_cm_s = 2.99792458e10
c_pc_yr = 0.30660139
yr2s = 31557600.0
```

```python
data = np.fromfile(file_final, dtype='=f8').reshape(-1, 4) # numpy calculate the number of elements in "-1" automatically
x, y, z, E = data.T
arrivals = np.fromfile(file_earth, dtype='=f8').reshape(-1, 2)
E_arr, time_arr = arrivals.T
N = len(E)
if N != expected_particles:
    raise ValueError('Particle count does not match the configured injection grid.')
if not np.all(np.isfinite(data)) or np.any(E <= m_p):
    raise ValueError('Final records contain invalid positions or energies.')
if not np.all(np.isfinite(arrivals)) or np.any(E_arr <= m_p) or np.any((time_arr < 0) | (time_arr > t_end)):
    raise ValueError('Arrival energies or times do not match the configured format/run.')
E_start = np.logspace(np.log10(E_min), np.log10(E_max), N)
T = E - m_p
print('Final particles:', N)
print('Final total energy min / median / max [GeV]:', np.percentile(E, [0, 50, 100]))
print('Solar-ring step samples:', len(arrivals))
if len(arrivals):
    print('Arrival total energy min / median / max [GeV]:', np.percentile(E_arr, [0, 50, 100]))
    print('Arrival time min / median / max [Myr]:', np.percentile(time_arr / 1e6, [0, 50, 100]))
```

## Continuous cooling reference
For total energy $E$ and kinetic energy $T=E-m_p$ (rest energy written as $m_p$ in GeV),
$$\dot E=-\frac{a}{3}\frac{E^2-m_p^2}{E}-\frac12n_H\sigma_{pp}(T)cT.$$
The reference integrates $\ln T$, so the reconstructed kinetic energy stays positive. Constant density and divergence make cooling independent of the random trajectory. Record order recovers each initial energy.


```python
def sigma_pp(kinetic):
    threshold = 0.2797  # GeV, proton kinetic-energy threshold
    # Evaluate the logarithm safely; the cross section is zero below threshold.
    safe = np.maximum(kinetic, threshold)
    log_ratio = np.log(safe / threshold)
    sigma = (30.7 - 0.96*log_ratio + 0.18*log_ratio**2)
    sigma = sigma * (1 - (threshold/safe)**1.9)**3 * 1e-27
    return np.where(kinetic > threshold, sigma, 0.0)


def cooling_rhs(q, logT):
    kinetic = np.exp(logT)
    total = kinetic + m_p
    dlogT_dt = -(a/3) * (total + m_p) / total
    dlogT_dt -= 0.5 * nH * sigma_pp(kinetic) * c_cm_s * yr2s
    # q=t/t_end, hence d(logT)/dq = t_end*d(logT)/dt.
    return t_end * dlogT_dt


# The energy-dependent pp cross section prevents using the adiabatic formula
# alone. solve_ivp adaptively integrates d(logT)/dq for all initial energies.
# Input: log kinetic energies at q=0. Output sol.y[:, -1]: values at q=1.
# rtol and atol control local numerical error, not the C++ splitting error.
sol = solve_ivp(cooling_rhs, (0, 1), np.log(E_start-m_p),
                rtol=1e-9, atol=1e-11)
if not sol.success:
    raise RuntimeError(sol.message)
E_reference = m_p + np.exp(sol.y[:, -1])
p_start = np.sqrt(E_start**2 - m_p**2)
p_adiabatic = p_start * np.exp(-a*t_end/3)
E_adiabatic = np.sqrt(p_adiabatic**2 + m_p**2)
residual = (E - E_reference) / E_reference
print('Max absolute relative cooling residual:', np.max(np.abs(residual)))
print('RMS relative cooling residual:', np.sqrt(np.mean(residual**2)))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].loglog(E_start, E, label='MC simulation')
axes[0].loglog(E_start, E_reference, '--', label='Combined-loss reference')
axes[0].loglog(E_start, E_adiabatic, ':', label='Adiabatic only')
axes[0].set_xlabel('Initial total energy [GeV]')
axes[0].set_ylabel('Final total energy [GeV]')
axes[0].legend()
axes[1].semilogx(E_start, residual)
axes[1].axhline(0, color='k', linestyle='--')
axes[1].set_xlabel('Initial total energy [GeV]')
axes[1].set_ylabel('(Recorded - reference) / reference')
fig.tight_layout()
plt.show()
```

## Final energy spectrum
Histogram the same discrete injection grid after theoretical cooling and compare with simulation using identical bins. Divide counts by bin widths to obtain densities; counts alone are not numerically equal to $E\,dN/dE$. Finite bins and endpoint placement can produce differences even for a small energy residual.


```python
bins = np.logspace(np.log10(min(E.min(), E_reference.min(), E_adiabatic.min())*0.999),
                   np.log10(E_max*1.001), 61)
energy_centers = np.sqrt(bins[:-1] * bins[1:])
dE = np.diff(bins)
dlnE = np.diff(np.log(bins))
simulation_counts = np.histogram(E, bins=bins)[0]
reference_counts = np.histogram(E_reference, bins=bins)[0]
adiabatic_counts = np.histogram(E_adiabatic, bins=bins)[0]
injection_counts = np.histogram(E_start, bins=bins)[0]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for i in range(2):
    if i == 0:
        width = dE
        ylabel = r'$dN/dE$ [particles GeV$^{-1}$]'
    else:
        width = dlnE
        ylabel = r'$E\,dN/dE$ [particles]'
    axes[i].step(energy_centers, simulation_counts/width, where='mid', label='MC simulation')
    axes[i].scatter(energy_centers, reference_counts/width, s=12, label='Combined-loss reference')
    axes[i].step(energy_centers, adiabatic_counts/width, where='mid', linestyle='-.', label='Adiabatic only')
    axes[i].step(energy_centers, injection_counts/width, where='mid', linestyle='--', label='Injection')
    axes[i].set_xscale('log')
    axes[i].set_yscale('log')
    axes[i].set_xlabel('Total proton energy [GeV]')
    axes[i].set_ylabel(ylabel)
    axes[i].legend()
fig.tight_layout()
plt.show()
```

## Final spatial distribution
All particles start at the origin. The vertical wind stretches the distribution in z, while x and y have no wind. There is no escape boundary. Positions are converted from pc to kpc below.


```python
R = np.sqrt(x**2 + y**2)
print('Mean position [kpc]:', np.mean(data[:, :3], axis=0)/1000)
print('Coordinate standard deviations [kpc]:', np.std(data[:, :3], axis=0)/1000)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(x/1000, y/1000, s=1, alpha=0.3)
angle = np.linspace(0, 2*np.pi, 300)
axes[0].plot(8*np.cos(angle), 8*np.sin(angle), 'r--', label='Solar ring')
axes[0].set_aspect('equal')
axes[0].set_xlabel('x [kpc]')
axes[0].set_ylabel('y [kpc]')
axes[0].legend()
axes[1].scatter(R/1000, z/1000, s=1, alpha=0.3)
axes[1].set_xlabel('R [kpc]')
axes[1].set_ylabel('z [kpc]')
fig.tight_layout()
plt.show()
```

## Solar-ring energy and time samples
Selection is $(R-8000\,\mathrm{pc})^2+z^2<\lambda(E)^2$, where $\lambda=2(\mathcal R/10\,\mathrm{GV})^{1/3}$ pc and $\mathcal R=\sqrt{E^2-m_p^2}$ in GV. Higher-energy samples use a larger selection volume and a longer time step. These are raw, correlated step samples, not unique arrivals or a flux-normalized spectrum. The files cannot recover first-arrival times or particle-by-particle uncertainty.


```python
arrival_counts = np.histogram(E_arr, bins=bins)[0]
time_bins = np.logspace(5, 9, 41)  # yr
sample_time_counts = np.histogram(time_arr, bins=time_bins)[0]
time_centers = np.sqrt(time_bins[:-1] * time_bins[1:])

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].step(energy_centers, arrival_counts/dlnE, where='mid')
axes[0].set_xscale('log')
if np.any(arrival_counts > 0):
    axes[0].set_yscale('log')
axes[0].set_xlabel('Arrival total energy [GeV]')
axes[0].set_ylabel(r'$dN_{sample}/d\ln E$ [step samples]')
axes[1].step(time_centers, sample_time_counts, where='mid')
axes[1].set_xlabel('CR proton age [yr]')
axes[1].set_ylabel('dN/dlnT')
axes[1].set_xscale("log")
axes[1].set_yscale("log")
fig.tight_layout()
plt.show()
```

## Interpretation
The continuous cooling reference uses a more accurate time integrator than the C++ first-order splitting, so small residuals include splitting and floating-point error. Quantify numerical convergence with a separate timestep refinement run.

Final particles all have the same propagation time. Solar-ring samples accumulate throughout the run, so their energy spectrum describes a different population and should not be normalized or interpreted as a final-state survival probability. No reweighting to a physical source spectrum is applied; injection is uniform in log total energy.
