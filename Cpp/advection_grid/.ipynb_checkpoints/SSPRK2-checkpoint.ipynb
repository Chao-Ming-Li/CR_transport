```python
from astropy import constants as const
from astropy import units as u
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

NR = 105
NZ = 105
NT = 100
Rf = np.linspace(-2,102,NR)
Zf = np.linspace(-2,102,NZ)
Rc = (Rf[1:] + Rf[:-1]) /2
Zc = (Zf[1:] + Zf[:-1]) /2
dV = Rc[:, None] * np.diff(Rf)[:,None] * np.diff(Zf)[None,:]
t = np.arange(100)*0.1
c_lst = ["red", "blue", "orange", "green", "purple", "yellow", "cyan", "magenta", "black", "pink", "lime"]
def numerical_diffusion(ndis):
    N_tot = np.sum(ndis * dV[None,:], axis = (1,2))
    mc_R = np.sum(ndis * Rc[None, :, None] * dV[None,:], axis = (1,2)) / N_tot
    mc_Z = np.sum(ndis * Zc[None, None, :] * dV[None,:], axis = (1,2)) / N_tot
    
    sigma_R = np.sum((Rc[None,:,None] - mc_R[:,None,None])**2 * ndis * dV[None,:], axis = (1,2)) / N_tot 
    sigma_Z = np.sum((Zc[None,None,:] - mc_Z[:,None,None])**2 * ndis * dV[None,:], axis = (1,2)) / N_tot 
    cov = np.sum((Rc[None,:,None]-mc_R[:,None,None])*(Zc[None,None,:]-mc_Z[:,None,None])*ndis*dV[None,:,:],axis=(1,2))/N_tot
    D_R = (sigma_R[1:] - sigma_R[0]) / 2 / t[1:] 
    D_Z = (sigma_Z[1:] - sigma_Z[0]) / 2 / t[1:] 

    return N_tot, mc_R, mc_Z, sigma_R, sigma_Z, cov, D_R, D_Z
t
```

```python
ndis_C_tvd_vF1 = np.fromfile("ndis_C_dt0.5_t200_vF1_tvd.bin",dtype = np.float64).reshape(NT, NR - 1, NZ - 1)
ndis_C_tvd_SSPRK2_vF1 = np.fromfile("ndis_C_dt0.5_t200_vF1_tvd_SSPRK2.bin",dtype = np.float64).reshape(NT, NR - 1, NZ - 1)
fig, ax = plt.subplots()

ndis_C_tvd_log = np.log10(ndis_C_tvd_vF1 + 1e-100)
ndis = ax.pcolormesh(Rc, Zc, ndis_C_tvd_log[0].T,shading="auto", vmin = -10, vmax = 1 )

cbar = fig.colorbar(ndis, ax=ax)
cbar.set_label('linear scale') 

def animate(i):
    ndis.set_array(ndis_C_tvd_log[i].T.ravel())
    return ndis,

ani = animation.FuncAnimation(
    fig,
    animate,
    frames=100,
    interval=10,
    blit=False  
)

plt.close(fig)
HTML(ani.to_html5_video())
```

```python
ndis_C_tvd = np.fromfile("ndis_C_dt0.1_t1000_vF_tvd.bin",dtype = np.float64).reshape(NT, NR - 1, NZ - 1)
ndis_C_tvd_SSPRK2 = np.fromfile("ndis_C_dt0.1_t1000_vF_tvd_SSPRK2.bin",dtype = np.float64).reshape(NT, NR - 1, NZ - 1)

fig, axes = plt.subplots(1,2, figsize=(12,5))
for i in range(10):
    axes[0].plot(Zc, ndis_C_tvd[i*10, :, 30], label = f" t = {i*100}", c = c_lst[i])
for i in range(10):
    axes[0].plot(Zc, ndis_C_tvd_SSPRK2[i*10,:, 30], label = f"SSPRK2",c = c_lst[i], linestyle="--")
for i in range(10):
    axes[1].plot(Zc, ndis_C_tvd[i*10, 30], label = f" t = {i*100}", c = c_lst[i])
for i in range(10):
    axes[1].plot(Zc, ndis_C_tvd_SSPRK2[i*10, 30], label = f"SSPRK2",c = c_lst[i], linestyle="--")

for i in range(2):
    axes[i].set_yscale("log")
    axes[i].set_ylim(1e-15,1e0)
    axes[i].legend(ncol=2)
axes[0].set_xlabel("R")
axes[1].set_xlabel("z")
```
