# CR diffusion and advection analysis
Analyze `ndis_C_v300_D1e29_cp.bin` using its matching `transport_grid.csv`.

The initial burst is $n(R,z,0)=\exp[-((R-10)^2+(z-10)^2)/9]$, with distances in kpc. The axisymmetric solver uses $D=10^{29}$ cm²/s, $v_z=300$ km/s, and evolves for $10^8$ yr without losses or interactions. Lower boundaries reflect; the outer boundaries at 200 kpc absorb.

Density has arbitrary amplitude. Integrals below use cylindrical volumes $\pi(R_{i+1}^2-R_i^2)\Delta z_j$. Free-space vertical references are approximate near the reflecting plane. Run cells in order; dependencies: NumPy and Matplotlib.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

# Set this directory explicitly if the kernel starts elsewhere.
DATA_DIR = Path.cwd()
BINARY = DATA_DIR / "ndis_C_v300_D1e29_cp.bin"
GRID_FILE = DATA_DIR / "transport_grid.csv"
DT_YR = 1e4
NT = 10000
SAVE_EVERY = 10
NG = 2
KPC_CM = 3.0856775814913673e21
YEAR_S = 31557600.0
D = 1e29 * YEAR_S / KPC_CM**2       # kpc^2/yr
VZ = 300.0 * 1e5 * YEAR_S / KPC_CM  # kpc/yr
plt.rcParams.update({"figure.dpi": 110, "axes.grid": False})
```

## Load geometry and snapshots
The native-endian float64 file has no header. Each snapshot includes two ghost cells on each side and uses `(R, z)` axis order. The loader checks the file length against the current writer's schedule, including its initial snapshot. A partial run is allowed only if it ends at a complete snapshot. Older files with a different schedule require matching metadata; their times cannot be inferred from the binary alone.

```python
geometry = np.atleast_1d(np.genfromtxt(GRID_FILE, delimiter=",", names=True))
R = geometry["R_centroid_kpc"]
z = geometry["z_center_kpc"]
dR = geometry["dR_kpc"]
dz = geometry["dz_kpc"]
NR, NZ = len(R), len(z)  # current writer uses identical radial/vertical axes
assert NR > 1 and np.all(dR > 0) and np.all(dz > 0)
R_faces = np.r_[0.0, np.cumsum(dR)]
z_faces = np.r_[0.0, np.cumsum(dz)]
assert np.allclose(z, 0.5 * (z_faces[:-1] + z_faces[1:]))
R_expected = (2.0 / 3.0) * (R_faces[1:]**2 + R_faces[1:] * R_faces[:-1]
                            + R_faces[:-1]**2) / (R_faces[1:] + R_faces[:-1])
assert np.allclose(R, R_expected), "Grid centroid/width mismatch"
volume = np.pi * np.diff(R_faces**2)[:, None] * dz[None, :]

snapshot_shape = (NR + 2 * NG, NZ + 2 * NG)
bytes_per_snapshot = np.prod(snapshot_shape) * np.dtype("=f8").itemsize
n_snapshots, remainder = divmod(BINARY.stat().st_size, int(bytes_per_snapshot))
if remainder or not n_snapshots:
    raise ValueError("Binary is empty or ends in a partial snapshot; check the matching grid")
saved_steps = np.unique(np.r_[0, np.arange(SAVE_EVERY, NT + 1, SAVE_EVERY), NT])
if n_snapshots > len(saved_steps):
    raise ValueError("Too many snapshots for this schedule; check for appended/legacy output")
time_yr = saved_steps[:n_snapshots] * DT_YR
time_myr = time_yr / 1e6
raw = np.memmap(BINARY, dtype="=f8", mode="r", shape=(n_snapshots, *snapshot_shape))
density = raw[:, NG:-NG, NG:-NG]
print(f"Grid: {NR} × {NZ}; snapshots: {n_snapshots}; last time: {time_myr[-1]:g} Myr")
print(f"Extent: R={R_faces[-1]:g}, z={z_faces[-1]:g} kpc; file={BINARY.stat().st_size / 2**20:.1f} MiB")
if n_snapshots < len(saved_steps):
    print("Partial run: diagnostics use only completed snapshots.")
expected_initial = np.exp(-((R[:, None] - 10)**2 + (z[None, :] - 10)**2) / 9)
if not np.allclose(density[0], expected_initial, rtol=1e-10, atol=1e-13):
    raise ValueError("First snapshot is not the expected t=0 source; check writer version/metadata")
```

## Density maps
Common logarithmic color limits show spreading and peak dilution. Only physical cells are plotted. Values at or below zero are masked here and checked separately below.

```python
selected = np.unique([np.argmin(abs(time_myr - t)) for t in (0, 10, 50, 100)])
peak_initial = float(density[0].max())
fig, axes = plt.subplots(1, len(selected), figsize=(4 * len(selected), 4),
                         squeeze=False, constrained_layout=True)
for ax, k in zip(axes.ravel(), selected):
    image = ax.pcolormesh(R_faces, z_faces, np.ma.masked_less_equal(density[k].T, 0),
                         norm=LogNorm(peak_initial * 1e-6, peak_initial),
                         cmap="magma", shading="flat")
    ax.set(xlabel="R [kpc]", ylabel="z [kpc]", title=f"t = {time_myr[k]:g} Myr",
           xlim=(0, min(70, R_faces[-1])), ylim=(0, min(90, z_faces[-1])))
fig.colorbar(image, ax=axes.ravel().tolist(), label="CR density [arbitrary units]")
plt.show()
```

## Conservation, positivity, and moments
Compute one snapshot at a time to keep memory use small. Moments use surviving CRs. Exact cell averages of $R^2$, $z$, and $z^2$ account for the finite cell volumes. Absorbing boundaries can reduce total number; significant growth or negative density indicates a numerical problem.

```python
R2_cell = 0.5 * (R_faces[:-1]**2 + R_faces[1:]**2)
z2_cell = (z_faces[:-1]**2 + z_faces[:-1] * z_faces[1:] + z_faces[1:]**2) / 3
mass = np.empty(n_snapshots)
mean_z = np.empty(n_snapshots)
var_z = np.empty(n_snapshots)
mean_R2 = np.empty(n_snapshots)
minimum = np.empty(n_snapshots)
for k, snapshot in enumerate(density):
    if not np.isfinite(snapshot).all():
        raise ValueError(f"Nonfinite density in snapshot {k}")
    weights = snapshot * volume
    mass[k] = weights.sum()
    if mass[k] <= 0:
        raise ValueError(f"Nonpositive total number in snapshot {k}")
    mean_z[k] = (weights * z[None, :]).sum() / mass[k]
    var_z[k] = (weights * z2_cell[None, :]).sum() / mass[k] - mean_z[k]**2
    mean_R2[k] = (weights * R2_cell[:, None]).sum() / mass[k]
    minimum[k] = snapshot.min()

reference_z = mean_z[0] + VZ * time_yr
reference_var_z = var_z[0] + 2 * D * time_yr
reference_R2 = mean_R2[0] + 4 * D * time_yr
fig, axes = plt.subplots(2, 2, figsize=(10, 7), constrained_layout=True)
axes[0, 0].plot(time_myr, mass / mass[0])
axes[0, 0].set(ylabel="Total number / initial number")
for ax, numerical, reference, label in [
    (axes[0, 1], mean_z, reference_z, "Mean z [kpc]"),
    (axes[1, 0], var_z, reference_var_z, "Vertical variance [kpc²]"),
    (axes[1, 1], mean_R2, reference_R2, "Mean R² [kpc²]"),
]:
    ax.plot(time_myr, numerical, label="Grid")
    ax.plot(time_myr, reference, "--", label="Free-space reference")
    ax.set(ylabel=label)
    ax.legend()
for ax in axes.ravel():
    ax.set_xlabel("Time [Myr]")
    ax.grid(alpha=0.25)
plt.show()
print(f"Final number / initial = {mass[-1] / mass[0]:.10f}")
print(f"Minimum physical density = {minimum.min():.6g}")
print(f"Largest relative number increase per snapshot = {np.max(np.diff(mass), initial=0) / mass[0]:.3e}")
print(f"Final mean z = {mean_z[-1]:.5f} kpc; reference = {reference_z[-1]:.5f}")
print(f"Final mean R² = {mean_R2[-1]:.5f} kpc²; reference = {reference_R2[-1]:.5f}")
```

## Volume-integrated radial and vertical profiles
These are probability densities in $dR$ and $dz$, normalized by the **initial** number. Their integrals equal the surviving fraction. The radial profile includes the cylindrical Jacobian and is not simply a cut through $n(R,z)$.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4), constrained_layout=True)
for k in selected:
    weights = density[k] * volume
    radial_profile = weights.sum(axis=1) / (mass[0] * dR)
    vertical_profile = weights.sum(axis=0) / (mass[0] * dz)
    axes[0].plot(R, radial_profile, label=f"{time_myr[k]:g} Myr")
    axes[1].plot(z, vertical_profile, label=f"{time_myr[k]:g} Myr")
axes[0].set(xlabel="R [kpc]", ylabel="Radial probability density [kpc⁻¹]", xlim=(0, 70))
axes[1].set(xlabel="z [kpc]", ylabel="Vertical probability density [kpc⁻¹]", xlim=(0, 90))
for ax in axes:
    ax.legend()
    ax.grid(alpha=0.25)
plt.show()
```

## Optional Monte Carlo comparison
Set `COMPARE_MC = True` only for `particles.bin` produced by the matching diffusion/advection run at $10^8$ yr. The headerless file cannot identify its own parameters. Positions are stored in pc, with one `(x,y,z)` triple per survivor. Counts are divided by the original 10000 particles, retaining any escape fraction.

```python
COMPARE_MC = False
MC_FILE = DATA_DIR / "../../Monte Carlo/particles.bin"
MC_INITIAL_COUNT = 10000
if COMPARE_MC:
    if not np.isclose(time_yr[-1], 1e8):
        raise ValueError("Monte Carlo comparison requires the final 1e8-year grid snapshot")
    positions = np.fromfile(MC_FILE, dtype="=f8")
    if positions.size % 3:
        raise ValueError("Expected native float64 xyz triples")
    positions = positions.reshape(-1, 3) / 1000.0
    if not np.isfinite(positions).all() or len(positions) > MC_INITIAL_COUNT:
        raise ValueError("Invalid particle file or wrong initial particle count")
    radius = np.hypot(positions[:, 0], positions[:, 1])
    # Coarser bins reduce sampling noise in this 10000-particle comparison.
    R_bins = np.linspace(0, R_faces[-1], 101)
    z_bins = np.linspace(0, z_faces[-1], 101)
    fig, axes = plt.subplots(1, 2, figsize=(10, 4), constrained_layout=True)
    final_weights = density[-1] * volume
    for ax, samples, bins, centers, widths, grid_counts, label in [
        (axes[0], radius, R_bins, R, dR, final_weights.sum(axis=1), "R"),
        (axes[1], positions[:, 2], z_bins, z, dz, final_weights.sum(axis=0), "z"),
    ]:
        counts, edges = np.histogram(samples, bins=bins)
        ax.stairs(counts / (MC_INITIAL_COUNT * np.diff(edges)), edges, label="Monte Carlo")
        ax.plot(centers, grid_counts / (mass[0] * widths), label="Grid")
        ax.set(xlabel=f"{label} [kpc]", ylabel="Probability density [kpc⁻¹]",
               xlim=(0, 70 if label == "R" else 90))
        ax.legend()
    plt.show()
    print(f"MC surviving fraction = {len(positions) / MC_INITIAL_COUNT:.6f}")
```
