```python
from astropy import constants as const
from astropy import units as u
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

def grid_R(NR, dR):
    R = np.zeros(NR)
    l = 10 / dR
    for i in range(NR):
        if i * dR < 10:
            R[i] = i * dR
        else:
            R[i] = 10** ((i + l ) / 2 / l)
    return R
def grid_Z(NZ, dZ):
    Z = np.zeros(NZ)
    l = 1 / dZ
    for i in range(NZ):
        if i * dZ < 1:
            Z[i] = i * dZ
        else:
            Z[i] = 10** ((i - l) / l)
    return Z
NR = 300
NZ = 150
Rf = grid_R(NR + 1,0.1)
Zf = grid_Z(NZ + 1,0.02)
Rc = (Rf[1:] + Rf[:-1]) / 2
Zc = (Zf[1:] + Zf[:-1]) / 2
```
