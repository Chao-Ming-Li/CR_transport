```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import random

fig, ax = plt.subplots()

N = 1000
x = np.zeros(N)
y = np.zeros(N)
l = 1

line, = ax.plot([], [], lw=1)

ax.set_xlim(-25, 25)
ax.set_ylim(-25, 25)

def animate(i):
    if i == 0:
        x[0], y[0] = 0, 0

    theta = random.uniform(-np.pi, np.pi)

    x[i] = x[i-1] + l * np.cos(theta)
    y[i] = y[i-1] + l * np.sin(theta)

    line.set_data(x[:i+1], y[:i+1])
    return line,

ani = animation.FuncAnimation(
    fig,
    animate,
    frames=300,
    interval=20,
    blit=True
)

plt.close(fig)
HTML(ani.to_html5_video())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

ndis_Pe1_gf_anime = np.fromfile(
    "ndis_Pe10_ci_TVD_2D_anime.bin",
    dtype=np.float64
).reshape(1001, 201, 201)

fig, ax = plt.subplots()

NX = 201
NY = 201

x = np.linspace(-100, 100, NX)
y = np.linspace(-100, 100, NY)

# 初始图
ndis = ax.pcolormesh(
    x, y,
    np.log(ndis_Pe1_gf_anime[0] + 1e-100).T,
    shading="auto",vmin=-10, vmax=5
)
cbar = fig.colorbar(ndis, ax=ax, label="number density log(n)")
title = ax.set_title("t = 0")

def animate(i):
    ndis.set_array(np.log(ndis_Pe1_gf_anime[i] + 1e-100).T.ravel())
    title.set_text(f"t = {i*100}")
    return ndis, title

ani = animation.FuncAnimation(
    fig,
    animate,
    frames=1000,
    interval=20,
    blit=False  
)

plt.close(fig)
HTML(ani.to_html5_video())
```
