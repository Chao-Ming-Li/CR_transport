```python
import numpy as np
from matplotlib import pyplot as plt
import matplotlib
matplotlib.rcParams["mathtext.fontset"] = "cm"
plt.rcParams["font.family"] = "serif"
import matplotlib.colors as colors
import gc
c_lst = ["red", "orange", "green", "blue", "cyan", "purple", "magenta"]
```

```python
NX = 20001
NT = 7
D = 1000
L = 10000.0
q = 1.0
v = 1.0
t = np.logspace(-4, 2, NT)
x = np.linspace(-L,L,NX)

ndis_Pe001_gf_cd = np.fromfile("ndis_Pe001_gf_cd.bin", dtype=np.float64)
ndis_Pe001_gf_cd = ndis_Pe001_gf_cd.reshape(NT,NX)
ndis_Pe001_gf_uw = np.fromfile("ndis_Pe001_gf_uw.bin", dtype=np.float64)
ndis_Pe001_gf_uw = ndis_Pe001_gf_uw.reshape(NT,NX)
ndis_Pe001_gf_tvd = np.fromfile("ndis_Pe001_gf_TVD.bin", dtype=np.float64)
ndis_Pe001_gf_tvd = ndis_Pe001_gf_tvd.reshape(NT,NX+3)

for i in range(NT):
    plt.plot(x, ndis_Pe001_gf_tvd[i,2:-1], label= "TVD t = 10^" + str(i-4), c = c_lst[i])
for i in range(NT):
    plt.scatter(x, ndis_Pe001_gf_cd[i], label= "cd", c = c_lst[i])
for i in range(NT):
    plt.plot(x, ndis_Pe001_gf_uw[i], label= "uw",linestyle="--", c = c_lst[i])
plt.xscale('symlog', linthresh=1)
plt.yscale("log")
plt.ylim(1e-5,1e0)
plt.legend(ncol=3,bbox_to_anchor=(1.01, 1), loc='upper left')
plt.xlabel("Distance")
plt.ylabel("n")
plt.title("Pe = 0.001")
```

```python
NX = 20001
NT = 7
D = 1000
L = 10000.0
q = 1.0
v = 1.0
t = np.logspace(-4, 2, NT)
x = np.linspace(-L,L,NX)
ndis_Pe001_ana = np.exp(-(x[None,:] - v*t[:,None])**2 / 4 / D / t[:,None]) / np.sqrt(4*np.pi*D*t[:,None])


fig, axes = plt.subplots(1,3,figsize=(15,5))
ndis_Pe001_gf_cd = np.fromfile("ndis_Pe001_gf_cd.bin", dtype=np.float64)
ndis_Pe001_gf_cd = ndis_Pe001_gf_cd.reshape(NT,NX)
ndis_Pe001_gf_uw = np.fromfile("ndis_Pe001_gf_uw.bin", dtype=np.float64)
ndis_Pe001_gf_uw = ndis_Pe001_gf_uw.reshape(NT,NX)
ndis_Pe001_gf_tvd = np.fromfile("ndis_Pe001_gf_TVD.bin", dtype=np.float64)
ndis_Pe001_gf_tvd = ndis_Pe001_gf_tvd.reshape(NT,NX+3)

steps = [1,10,100]
for i in range(3):
    axes[i].plot(x, ndis_Pe001_ana[i*3], label= "Analytical t = 10^" + str(i*3-4), c = c_lst[i])
    axes[i].plot(x, ndis_Pe001_gf_tvd[i*3,2:-1], label= "TVD ", linestyle="--", c = c_lst[i])
    axes[i].scatter(x[::steps[i]], ndis_Pe001_gf_cd[i*3,::steps[i]], label= "cd", marker = "s", edgecolor=c_lst[i],facecolor="none")
    axes[i].scatter(x[::steps[i]], ndis_Pe001_gf_uw[i*3,::steps[i]], label= "uw", marker = "x", c=c_lst[i])
    axes[i].set_xscale('symlog', linthresh=1)
    axes[i].set_yscale("log")
    axes[i].set_ylim(1e-5,1e0)
    axes[i].set_xlim(-2e3,2e3)

    # axes[i].legend(ncol=3,bbox_to_anchor=(1.01, 1), loc='upper left')
    axes[i].set_xlabel("Distance")
    axes[i].set_ylabel("n")
    axes[i].legend()
fig.suptitle("Pe = 0.001", fontsize=16)
```
