```python
import numpy as np
from matplotlib import pyplot as plt
import matplotlib
matplotlib.rcParams["mathtext.fontset"] = "cm"
plt.rcParams["font.family"] = "serif"
import matplotlib.colors as colors
import gc
```

```python
NX = 20001
NT = 7
D = 1000.0
L = 10000.0
q = 1.0
v = 1.0
t = np.logspace(-4, 2, NT)
x = np.linspace(-L,L,NX)

ndis_Pe001_gf_cd = np.fromfile("ndis_Pe001_gf_cd.bin", dtype=np.float64)
ndis_Pe001_gf_cd = ndis_Pe001_gf_cd.reshape(NT,NX)
ndis_Pe001_gf_uw = np.fromfile("ndis_Pe001_gf_uw.bin", dtype=np.float64)
ndis_Pe001_gf_uw = ndis_Pe001_gf_uw.reshape(NT,NX)
ndis_Pe001_ana = np.exp(-(x[None,:] - v*t[:,None])**2 / 4 / D / t[:,None]) / np.sqrt(4*np.pi*D*t[:,None])

fig, axes = plt.subplots(1,2, figsize=(12,5))
for i in range(NT):
    axes[0].plot(x, ndis_Pe001_gf_cd[i], label="t = 10^" + str(i-4) + " cd")
for i in range(NT):
    axes[0].scatter(x, ndis_Pe001_gf_uw[i], label="up")

for i in range(NT):
    axes[1].plot(x, ndis_Pe001_gf_cd[i], label= " cd")
for i in range(7):
    axes[1].scatter(x, ndis_Pe001_ana[i], label="ana")

for i in range(2):
    axes[i].set_xscale('symlog', linthresh=1)  
    axes[i].set_yscale("log")
    axes[i].legend(ncol=2)
    axes[i].set_ylim(1e-5,1e1)
    axes[i].set_title("Pe = 0.001")
```
