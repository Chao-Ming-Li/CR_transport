# Spallation test
Analyze `spallation_test.bin` from `spallation.cpp`. Run this notebook from the simulation directory.

Each record contains four native-endian float64 values: x, y, z [pc] and total energy per nucleon E [GeV/n]. **Every initial particle is written**, including particles removed into untracked fragments. Species and propagation times are not saved. Consequently, measured isotope abundances, secondary spectra, and B/C cannot be recovered from this file.

The cells below examine the saved sample and calculate a separate theoretical prediction for the single-residual fragmentation network. No changes to the simulation are required.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
from scipy.linalg import expm
from matplotlib.colors import LogNorm

plt.rcParams.update({'figure.dpi': 120, 'axes.grid': False})
```

```python
# Parameters in the current spallation.cpp; the binary has no run metadata.
# Change these if the file was generated with different parameters.
file = Path("spallation_test.bin")
N = 21000
E_min = 10.0       # total energy per nucleon, GeV/n
E_max = 1e6
nH = 0.1          # cm^-3; constant in this run
t_end = 1e8        # yr
v_z = 3e-4        # pc / yr
c = 2.99792458e10  # cm / s
c_pc = 0.30660139  # pc / yr
yr2s = 31557600.0

if file.stat().st_size == 0 or file.stat().st_size % 32:
    raise ValueError("Expected nonempty records of four float64 values")
data = np.fromfile(file, dtype="=f8").reshape(-1, 4)
x, y, z, E = data.T
xyz = data[:, :3]
n = len(E)
assert np.isfinite(data).all() and np.all(E > 0)
assert n == N, "Particle count does not match the run parameters"

print("Number of records:", n)
print("E min / median / max [GeV/n]:", np.percentile(E, [0, 50, 100]))
print("Mean position [pc]:", xyz.mean(axis=0))
print("Coordinate standard deviations [pc]:", xyz.std(axis=0, ddof=1))
```

## Energy initialization
The source initializes a deterministic logarithmic energy grid and conserves energy per nucleon during fragmentation. Because removed records are retained, the saved energy distribution should equal the injected grid. This is an all-record spectrum, not a surviving-CR spectrum.

```python
Ei = np.logspace(np.log10(E_min), np.log10(E_max), n)
err = np.max(np.abs(E / Ei - 1))
print("Maximum relative difference from initial grid:", err)
assert np.allclose(E, Ei, rtol=1e-12, atol=0)

bins = np.logspace(np.log10(E_min), np.log10(E_max), 26)
bins[0] = np.nextafter(min(bins[0], E.min()), 0)
bins[-1] = np.nextafter(max(bins[-1], E.max()), np.inf)
counts, _ = np.histogram(E, bins=bins)
dE = np.diff(bins)
dlnE = np.diff(np.log(bins))
Ec = np.sqrt(bins[:-1] * bins[1:])
assert counts.sum() == n

plt.plot(Ec, counts / dlnE, "o-", label="All saved records")
plt.axhline(n / np.log(E_max / E_min), linestyle="--", color="black",
            label="Log-uniform continuum limit")
plt.xscale("log")
plt.xlabel("Total energy per nucleon [GeV/n]")
plt.ylabel(r"$dN/d\ln E$ [particles]")
plt.legend()
plt.show()
```

## Network prediction (not measured abundances)
Read the isotope order and cross-section matrix from `cross_section.hpp`. For a row vector of populations, the generator has off-diagonal entries nH c sigma(i → j), diagonal entries −nH c sigma(inelastic), and one absorbing state for untracked fragments. The matrix exponential includes all sequential fragmentation chains. This reproduces the constant-density, v ≈ c model in the source; it does not add radioactive decay or helium targets.

```python
header = Path("cross_section.hpp").read_text()
nuclide_block = header.split("nuclides", 1)[1].split("constexpr std::size_t index", 1)[0]
names = re.findall(r'\{\s*"([A-Za-z0-9]+)"\s*,', nuclide_block)
sigma_block = header.split("sigma{", 1)[1].split(";", 1)[0]
values = re.findall(r'(?<![\w.])[+-]?(?:\d+\.\d*|\d*\.\d+|\d+)(?:[eE][+-]?\d+)?', sigma_block)
k = len(names)
sigma = np.array(values, dtype=float).reshape(k, k)  # cm^2
assert k == 10
partial = sigma.copy()
np.fill_diagonal(partial, 0)
missing = np.diag(sigma) - partial.sum(axis=1)
assert np.all(partial >= 0) and np.all(missing >= 0)

rate = nH * c * yr2s
Q = np.zeros((k + 1, k + 1))
Q[:k, :k] = rate * partial
Q[np.arange(k), np.arange(k)] = -rate * np.diag(sigma)
Q[:k, k] = rate * missing
assert np.allclose(Q.sum(axis=1), 0, atol=1e-20)

initial = np.zeros(k + 1)
initial[names.index("O16")] = n * 10 / 21
initial[names.index("N14")] = n / 21
initial[names.index("C12")] = n * 10 / 21
pred = initial @ expm(Q * t_end)
assert np.all(pred >= -1e-8) and np.isclose(pred.sum(), n)

print("species       initial       predicted final")
for name, a, b in zip(names + ["untracked"], initial, pred):
    print(f"{name:>9} {a:13.0f} {b:21.6g}")
print("Predicted tracked fraction:", pred[:k].sum() / n)
C = pred[names.index("C12")] + pred[names.index("C13")]
B = pred[names.index("B10")] + pred[names.index("B11")]
print("Predicted B/C:", B / C if C > 0 else np.nan)
```

```python
times = np.r_[0, np.geomspace(1e3, t_end, 150)]
pop = np.array([initial @ expm(Q * t) for t in times])
for j, name in enumerate(names):
    plt.plot(times, pop[:, j] / n, label=name)
plt.plot(times, pop[:, -1] / n, "k--", label="Untracked")
plt.xscale("symlog", linthresh=1e3)
plt.yscale("log")
plt.xlabel("Propagation time [yr]")
plt.ylabel("Predicted fraction of initial particles")
plt.legend(ncol=3)
plt.show()
```

## Spatial distribution of all records
Removed particles stop at their removal step; tracked particles propagate to t_end. The plots therefore mix different propagation durations. The wind position v_z t_end is shown only as a reference for full-duration particles. It must not be used as the center of the whole sample for a diffusion estimate.

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4), constrained_layout=True)
for j, ax in enumerate(axes):
    ax.hist(xyz[:, j] / 1000, bins=80, alpha=0.7)
    ax.set(xlabel="xyz"[j] + " [kpc]", ylabel="Records / bin")
axes[2].axvline(v_z * t_end / 1000, color="black", linestyle="--",
                 label="Full-duration wind displacement")
axes[2].legend(fontsize=8)
plt.show()

fig, axes = plt.subplots(1, 3, figsize=(15, 4), constrained_layout=True)
for ax, (i, j) in zip(axes, [(0, 1), (0, 2), (1, 2)]):
    hist = ax.hist2d(xyz[:, i] / 1000, xyz[:, j] / 1000,
                     bins=80, norm=LogNorm(), cmin=1)
    ax.set(xlabel="xyz"[i] + " [kpc]", ylabel="xyz"[j] + " [kpc]", aspect="equal")
    fig.colorbar(hist[3], ax=ax, label="Records / bin")
plt.show()
```

## Mean stopping time and transverse diffusion
For constant density and energy-independent cross sections, the theoretical mean duration is the integral of the tracked survival fraction. Removed particles are moved through their entire final transport step before removal, so the saved mean duration slightly exceeds the continuous reaction-time prediction.

The transverse second moment avoids the varying vertical drift: ⟨x²+y²⟩ ≈ 4 D(E) ⟨t_stop⟩ with D = c_pc lambda / 3. Predictions below are averages over the actual energies in each bin. They test the all-record mixture, not diffusion of only the surviving nuclei.

```python
# Augment the generator with an accumulator for integrated tracked survival.
A = np.zeros((k + 2, k + 2))
A[:k + 1, :k + 1] = Q
A[:k, -1] = 1.0
p0 = np.r_[initial / n, 0.0]
mean_time = (p0 @ expm(A * t_end))[-1]
print("Predicted mean propagation duration [yr]:", mean_time)
print("Mean z / v_z [yr]:", z.mean() / v_z)
print("Predicted mean z [pc]:", v_z * mean_time)
print("Largest transport timestep [yr]:", 2 * E.max()**(1 / 3) / (2 * c_pc))

rows = []
for lower, upper in zip(bins[:-1], bins[1:]):
    selected = (E >= lower) & (E < upper)
    count = selected.sum()
    if count < 30:
        continue
    transverse = x[selected]**2 + y[selected]**2
    D = c_pc * 2 * E[selected]**(1 / 3) / 3
    rows.append((np.exp(np.log(E[selected]).mean()), transverse.mean(),
                 transverse.std(ddof=1) / np.sqrt(count), 4 * D.mean() * mean_time))
result = np.array(rows)
e_bin, measured, error, expected = result.T

plt.errorbar(e_bin, measured, yerr=error, fmt="o", label="Saved records")
plt.plot(e_bin, expected, label="Network + diffusion prediction")
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Total energy per nucleon [GeV/n]")
plt.ylabel(r"$\langle x^2+y^2\rangle$ [pc$^2$]")
plt.legend()
plt.show()

plt.errorbar(e_bin, measured / expected, yerr=error / expected, fmt="o")
plt.axhline(1, color="black", linestyle="--")
plt.xscale("log")
plt.xlabel("Total energy per nucleon [GeV/n]")
plt.ylabel("Measured / predicted transverse second moment")
plt.show()
```

## Information needed for measured spallation spectra
To measure isotope abundances, spectra, or B/C, save the final species ID for each binary record (including an untracked flag). Saving t[i] also permits diffusion checks using each particle's propagation duration. Initial species IDs would allow parent-to-daughter transition counts.

The theoretical populations above are expectations from the current header and assumed run parameters, not measurements from the binary. The source/header may have changed since the run. Energy initialization checks and spatial comparisons cannot establish the individual daughter outcomes.
